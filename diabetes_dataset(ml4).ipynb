```python
import pandas as pd
```

```python
pd.set_option('display.max_?columns', None)
```

```python
df = pd.read_csv('/content/diabetes_dataset.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.drop(['diabetes_stage'], axis=1, inplace=True)
```

```python
X = df.drop(['diagnosed_diabetes'], axis=1)
y = df['diagnosed_diabetes'].copy()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
import numpy as np
```

```python
num_features = X_train.select_dtypes(include=[np.number]).columns
cat_features = X_train.select_dtypes(exclude=[np.number]).columns
```

```python
num_pipeline = Pipeline([
    ('impute', SimpleImputer(strategy='median')),
    ('scale', StandardScaler())
])

cat_pipeline = Pipeline([
    ('impute', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(sparse_output= False))
])

transformer = ColumnTransformer([
    ('cat', cat_pipeline, cat_features),
    ('num', num_pipeline, num_features)
], remainder = 'passthrough')
```

```python
model = RandomForestClassifier
full_pipeline = Pipeline([
    ('preprocessing', transformer),
    ('estimator', model)
])
```

```python
model = RandomForestClassifier()
full_pipeline = Pipeline([
    ('preprocessing', transformer),
    ('estimator', model)
])
full_pipeline.fit(X_train, y_train)
```

```python
full_pipeline.score(X_train, y_train)
```

```python
full_pipeline.score(X_test, y_test)
```

```python
param_grid = {
    'estimator__n_estimators': [75, 100, 125],
    'estimator__criterion': ['gini', 'entropy', 'log_loss'],
    'estimator__n_jobs': [-1]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
gscv = GridSearchCV(full_pipeline, param_grid = param_grid, cv=3)
```

```python
gscv.fit(X_train, y_train)
```

```python
gscv.best_params_
```

```python
best_model = gscv.best_estimator_
```

```python
best_model.fit(X_train, y_train)
```

```python
best_model.score(X_train, y_train)
```

```python
best_model.score(X_test, y_test)
```
